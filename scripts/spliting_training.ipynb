```python
import pandas as pd

# Columns representing time order
time_cols = ["year", "month", "day", "hour"]

for i in range(1, 8):
    file_name = f"cleaned_site_{i}_train_data.csv"   # use cleaned files
    print(f"Splitting {file_name} ...")
    
    # Load data
    df = pd.read_csv(file_name)

    # Sort chronologically
    df = df.sort_values(time_cols)
    
    # Compute split indices
    n = len(df)
    train_end = int(0.70 * n)
    test_end = int(0.85 * n)

    # Create splits (no shuffling!)
    train_df = df.iloc[:train_end]
    test_df = df.iloc[train_end:test_end]
    val_df = df.iloc[test_end:]

    # Save outputs
    train_df.to_csv(f"site_{i}_train.csv", index=False)
    test_df.to_csv(f"site_{i}_test.csv", index=False)
    val_df.to_csv(f"site_{i}_val.csv", index=False)

    print(f"✔ Done: site_{i}_train.csv, site_{i}_test.csv, site_{i}_val.csv\n")
```
