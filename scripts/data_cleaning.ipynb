```python
import pandas as pd
```

```python


# List of satellite columns to clean
cols_to_clean = ["NO2_satellite", "HCHO_satellite", "ratio_satellite"]

# Loop through all 7 sites
for i in range(1, 8):
    file_name = f"site_{i}_train_data.csv"
    
    print(f"Processing {file_name} ...")
    
    # Load CSV
    df = pd.read_csv(file_name)
    
    # ---- Add missing flags BEFORE filling ----
    for col in cols_to_clean:
        df[col + "_missing"] = df[col].isna().astype(int)
    
    # ---- Forward fill + backward fill ----
    for col in cols_to_clean:
        df[col] = df[col].fillna(method="ffill")
        df[col] = df[col].fillna(method="bfill")
    
    # ---- Save cleaned data ----
    df.to_csv(f"cleaned_{file_name}", index=False)
    
    print(f"✔ Finished cleaning {file_name} → saved as cleaned_{file_name}\n")
```

# DATA VIZ.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pandas.plotting import scatter_matrix

# -------- Load any site file --------
df = pd.read_csv("site_1_train_data.csv")   # change to your file

# Ensure numeric columns are converted properly
df = df.apply(pd.to_numeric, errors="ignore")


# ============================================================
# 1) TIME-SERIES PLOT (first 500 hours)
# ============================================================

plt.figure(figsize=(12, 6))
n = min(500, len(df))

plt.plot(np.arange(n), df["O3_target"].iloc[:n], label="O3_target")
plt.plot(np.arange(n), df["NO2_target"].iloc[:n], label="NO2_target")

plt.title("Time-Series of O3 and NO2 (first 500 hours)")
plt.xlabel("Hour Index")
plt.ylabel("Concentration (μg/m³)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()


# ============================================================
# 2) CORRELATION HEATMAP (inputs + targets)
# ============================================================

corr_cols = [
    "O3_forecast", "NO2_forecast", "T_forecast", "q_forecast",
    "u_forecast", "v_forecast", "w_forecast",
    "NO2_satellite", "HCHO_satellite", "ratio_satellite",
    "O3_target", "NO2_target"
]

corr = df[corr_cols].corr()

plt.figure(figsize=(10, 8))
plt.imshow(corr, interpolation="nearest", aspect="auto")
plt.colorbar(label="Correlation (Pearson r)")

plt.xticks(range(len(corr_cols)), corr_cols, rotation=45, ha="right")
plt.yticks(range(len(corr_cols)), corr_cols)

# annotate correlation values
for i in range(len(corr_cols)):
    for j in range(len(corr_cols)):
        plt.text(j, i, f"{corr.values[i,j]:.2f}",
                 ha="center", va="center", fontsize=7)

plt.title("Correlation Heatmap: Inputs and Targets")
plt.tight_layout()
plt.show()


# ============================================================
# 3) SCATTER-MATRIX (pairwise)
# ============================================================

sm_cols = [
    "O3_forecast", "NO2_forecast", "T_forecast",
    "O3_target", "NO2_target"
]

# sample smaller subset for readability
df_small = df[sm_cols].dropna().sample(min(500, df[sm_cols].dropna().shape[0]), random_state=42)

scatter_matrix(df_small, figsize=(10, 10), diagonal="hist")
plt.suptitle("Scatter-Matrix of Key Variables", y=1.02)
plt.tight_layout()
plt.show()
```
